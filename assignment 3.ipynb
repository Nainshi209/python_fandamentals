```python
def two_sum(arr,target):
    left = 0
    right = len(arr)-1
    while left<right:
        total = arr[left]+arr[right]
        if total == target:
          return [arr[left],arr[right]]
        elif total<target:
          left+=1
        else:
          right-=1
    return None
arr =[2,7,12,8]
target = 9
print(two_sum(arr,target))
```

```python
def reverse_array(arr):
    left =0
    right = len(arr)-1
    while left<right:
        arr[left],arr[right]=arr[right],arr[left]
        left+=1
        right-=1
    return arr
arr = [1,3,5,11]
print(reverse_array(arr))
```

```python
def arrange_negative_positive(arr):
    left = 0
    right = len(arr)-1
    while left <right:

        if arr[left]<0:
            left+=1
        elif arr[right]>=0:
            right-=1
        else:
            arr[left],arr[right]=arr[right],arr[left]
            left+=1
            right-=1
    return arr
arr = [1,-2,3,4,-5,6,]
print(arrange_negative_positive(arr))
```

```python
def remove_duplicates(arr):
    if len(arr) ==0:
        return 0
    left = 0
    for right in range (1,len(arr)):
        if arr[right]!=arr[left]:
            left+=1
            arr[left]=arr[right]
    return left+1
arr = [1,1,2,2,3,3,4]
n = remove_duplicates(arr)
print(arr[:n])
print(n)

    
```

```python
def factorial(n):
    if n==0 or n==1:
        return 1
    else:
        return n*factorial(n-1)
n = int(input("enter the number:"))
print("factorial=",factorial(n))
```

```python
def isPrime(n):
    if n<2:
        return False
    for i in range(2,n):
        if n%i==0:
         return False
    return True
for n in range(1,51):
    if isPrime(n):
        print(n)
```

```python
def swap(a,b):
    a,b=b,a
    return a,b
a= int(input("enter first number:"))
b= int(input("enter second number:"))
a,b=swap(a,b)
print("after swaping:")
print("a=",a)
print("b=",b)
```

```python
def gcd(a,b):
    while b!=0:
        a,b=b,a%b
    return a
a=int(input("enter first number:"))
b=int(input("enter second number:"))
print("gcd=",gcd(a,b))
```

```python
def average(arr,size):
    total = 0
    for i in range(size):
        total = total + arr[i]
    return total/size
arr=[10,20,30,40,50,60]
size= len(arr)
print("average=",average(arr,size))
```
