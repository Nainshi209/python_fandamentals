```python
queue = []

#enqueue
queue.append(23)
queue.append(34)
queue.append(45)

print("Queue:", queue)

#dequeue
item = queue.pop(0)

print("removed:", item)
print("Queue:", queue)

#peak
print("front:", queue[0])

#size
print("size:", len(queue))

#empty
print("is empty:", len(queue) == 0)
```
