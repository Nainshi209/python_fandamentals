```python
def is_prime(n):
    if n<=1:
        return False
    
    for i in range(2, int(n**0.5) + 1):
        if n%i==0:
            return False
    return True


for num in range(1,31):
        if is_prime(num):
            print(num)
```

```python
def reverse_and_slice(text):
    reversed_text = text[::-1]
    first_half = text[:len(text)//2]
    
    return reversed_text, first_half


# Example
result = reverse_and_slice("Python")

print("Reversed string:", result[0])
print("First half:", result[1])

```
