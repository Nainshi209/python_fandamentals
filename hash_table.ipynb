```python
#given array 
arr= [30,34,32,55,60]

#size of our hash table 
size = 10

#create an empty hash table
hash_table=[None]*size

#go through every  number in the array
for num in arr:

    #caculate the index using hash function 
    index = num % size

    #store the number at calculate index
    hash_table[index]= num

    #print the hash table
print(hash_table)
```
