```python
def is_even(num):
    if num%2==0:
        return True
    else:
        return False
print(is_even(12))
print(is_even(9))
```

```python
def sum(a,b):
    return a+b
print(sum(10,20))
```

```python
def find_max(a,b,c):
    if a>=b and a>=c:
        return a
    elif b>=a and b>=c:
        return b
    else:
        return c
print(find_max(12,34,13))
```

```python
def count_vowels(word):
    count = 0
    for i in word:
        if i in "aeiouAEIOU":
            count = count+1
    return count
print(count_vowels("nainshi"))
```
