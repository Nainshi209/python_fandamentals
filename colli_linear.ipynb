```python
arr = [30,34,32,55,60,44,54]
size = 10
hash_table = [None]*size
for ele in arr:
    index = ele % size 
    #if index is already occupied
    while hash_table[index] is not None:
        #move the next index 
        index = (index+1 % size)

        # store the number at empty index
    hash_table[index] = ele
print(hash_table)
```

```python
#quadratic hashing

arr = [30,34,32,35,60,44,45]
size= 10
hash_table = [None]*size
for num in arr:
    index = num % size
    i=1
    while hash_table[index] is not None:
        index = (num  % size +1 *i) % size 
        i+=1
        break
    hash_table[index]= num
print(hash_table)
```
